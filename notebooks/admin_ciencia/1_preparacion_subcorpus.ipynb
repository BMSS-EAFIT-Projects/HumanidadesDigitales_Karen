```python
import pandas as pd
```

```python
ruta_corpus = "../../data/processed/corpus_date_formatted.xlsx"

corpus_completo = pd.read_excel(ruta_corpus, engine="openpyxl")

corpus_completo.head()
```

```python
ruta_id_doc_admin = "../../data/auxiliar/id_doc_admin.txt"

with open(ruta_id_doc_admin, "r", encoding="utf-8") as f:
    id_doc_admin = [line.strip() for line in f if line.strip()]

id_doc_admin[:20]
```

```python
id_doc_admin = id_doc_admin[1:]
id_doc_admin_int = [int(x) for x in id_doc_admin]

corpus_filtrado = corpus_completo[corpus_completo["ID"].isin(id_doc_admin_int)].copy()
corpus_filtrado.head()
```

```python
ruta_salida = "../../data/columnas_admin_ciencia.xlsx"
corpus_filtrado.to_excel(ruta_salida, index=False, engine="openpyxl")
```
